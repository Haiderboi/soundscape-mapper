## 1. Synthetic demo waveform

Replace `AUDIO_FILE` with a real recording path.  Until then we use a
synthetic signal composed of bird-like chirps.

```python
SR = 22050
t = np.linspace(0, 4, SR * 4, endpoint=False)

# Simulated bird chirp (AM-modulated sine)
chirp = 0.5 * np.sin(2 * np.pi * 3500 * t) * (0.5 + 0.5 * np.sin(2 * np.pi * 3 * t))
# Add low-frequency geophony (wind)
wind = 0.1 * np.random.default_rng(0).standard_normal(len(t))
y = (chirp + wind).astype(np.float32)

print(f'Synthetic signal: {len(y)} samples  sr={SR}')
```

## 2. Waveform & Spectrogram

```python
fig = plot_waveform(y, SR, title='Synthetic Soundscape – Waveform')
plt.show()
```

```python
fig = plot_spectrogram(y, SR, title='Log-Power Spectrogram')
plt.show()
```

```python
fig = plot_mel_spectrogram(y, SR, title='Mel Spectrogram')
plt.show()
```

```python
fig = plot_mfcc(y, SR, title='MFCCs')
plt.show()
```

## 3. Ecoacoustic Indices

```python
indices = compute_all_indices(y, SR)
for key, val in indices.items():
    print(f'  {key:20s}: {val:.4f}')
```

## 4. Spectral Features

```python
features = extract_all_features(y, SR)
print('Feature summary:')
for key in ['spectral_centroid_mean', 'spectral_bandwidth_mean',
            'spectral_rolloff_mean', 'zcr_mean', 'rms_mean']:
    print(f'  {key:30s}: {features[key]:.4f}')

print(f'\n  MFCC mean shape : {features["mfcc_mean"].shape}')
print(f'  Chroma mean shape: {features["chroma_mean"].shape}')
```
